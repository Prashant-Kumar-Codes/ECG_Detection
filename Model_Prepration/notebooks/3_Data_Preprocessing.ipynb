```python
from pathlib import Path

import pandas as pd
import numpy as np
```

```python
curr_path = Path.cwd()
Output_Datasets_path = curr_path.parent.parent / 'Output_Datasets'
```

```python
# data = np.load(wfdb_array_path)
```

```python
X_train_path = Output_Datasets_path / 'X_train.npy'
Y_train_path = Output_Datasets_path / 'Y_train.npy'
X_val_path = Output_Datasets_path / 'X_val.npy'
Y_val_path = Output_Datasets_path / 'Y_val.npy'
```

```python
print("Loading X_train and Y_train...")
X_train = np.load(X_train_path, allow_pickle=True)
Y_train = np.load(Y_train_path, allow_pickle=True)
print("Successfully loaded X_train and Y_train")
```

```python
print("Loading X_val and Y_val...")
X_val = np.load(X_val_path, allow_pickle=True)
Y_val = np.load(Y_val_path, allow_pickle=True)
print("Successfully loaded X_val and Y_val")
```

## Converting Mulit-Labels of Targets to Binary Vectors

```python
from sklearn.preprocessing import MultiLabelBinarizer
```

```python
label2Binary = MultiLabelBinarizer()
Y_train_bin = label2Binary.fit_transform(Y_train)
Y_val_bin = label2Binary.fit_transform(Y_val)
```

```python
print(f'Label classes: ', label2Binary.classes_)
print(f'Shape of Y_train_bin: {Y_train_bin.shape}')
print(f'Actual Labels: {Y_train[0]} \t Labels to binary vectors: {Y_train_bin[0]}')
print(f'Actual Labels: {Y_train[29]} \t Labels to binary vectors: {Y_train_bin[29]}')
```

## Scale/Normalize Signals

```python
from sklearn.preprocessing import StandardScaler
```

```python
# reshape inputs before scaling
num_rows, num_timestamps, num_leads = X_train.shape
X_train_flat = X_train.reshape(-1, num_leads)
X_val_flat = X_val.reshape(-1, num_leads)
```

```python
scaler = StandardScaler()
X_train_norm = scaler.fit_transform(X_train_flat).reshape(-1, num_timestamps, num_leads)
X_val_norm = scaler.fit_transform(X_val_flat).reshape(-1, num_timestamps, num_leads)
```

## Filter data having no Labels(superclass)

```python
# mask for filter
mask_train = np.sum(Y_train_bin, axis=1) > 0
mask_val = np.sum(Y_train_bin, axis=1) > 0
```

```python
print("Before filter")
print(f'Shape of X_train_norm: {X_train_norm.shape}')
print(f'Shape of Y_train_bin: {Y_train_bin.shape}')
```

```python
X_train_clean = X_train_norm[mask_train]
Y_train_clean = Y_train_bin[mask_train]
X_val_clean = X_train_norm[mask_val]
Y_val_clean = Y_train_bin[mask_val]
```

```python
print("After filter")
print(f'Shape of X_train_clean: {X_train_clean.shape}')
print(f'Shape of Y_train_clean: {Y_train_clean.shape}')
```

## Saving the data

```python
print('Saving X_train_clean and Y_train_clean...')
np.save(rf'{Output_Datasets_path}/X_train_clean', X_train_clean)
np.save(rf'{Output_Datasets_path}/Y_train_clean', Y_train_clean)
print('Successfully saved X_train_clean and Y_train_clean')
```

```python
print('Saving X_val_clean and Y_val_clean...')
np.save(rf'{Output_Datasets_path}/X_val_clean', X_val_clean)
np.save(rf'{Output_Datasets_path}/Y_val_clean', Y_val_clean)
print('Successfully saved X_val_clean and Y_val_clean')
```
