## 3. PyTorch DataSet 

```python
class EcgDataset(Dataset):
    def __init__(self, inputs, target):
        self.inputs = inputs
        self.target = target

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        return torch.tensor(inputs[idx], dtype=torch.float64), torch.tensor(target[idx], dtype=torch.float64)
```

```python
print('Preparing dataset...')
train_dataset = EcgDataset(inputs=X_train_clean, target=Y_train_clean)
val_dataset = EcgDataset(X_val_clean, Y_val_clean)
print('Completed dataset preparation!')
```
